```python
import omero_analysis_notebook as oan
ctx = oan.configure(r'''{"schema":"nl.bioimaging.omero-analysis-notebook.v1","inputs":[{"id":"measurements","kind":"query","path":"input/mini.csv","formats":["csv"],"required":true}],"results":{"path":"results"},"parameters":[{"name":"group","type":"choice","default":"control","choices":["control","treated"]}],"requirements":["pandas"]}''')
ctx.display_parameters()
```

```python
summary = await ctx.query("measurements", "SELECT category, COUNT(*) AS count, AVG(value) AS mean_value FROM data WHERE category = $group GROUP BY category", {"group": ctx.params["group"]})
summary.to_csv(ctx.results / "csv-summary.csv", index=False)
summary
```
