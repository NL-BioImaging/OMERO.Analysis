```python
import omero_analysis_notebook as oan
ctx = oan.configure(r'''{"schema":"nl.bioimaging.omero-analysis-notebook.v1","inputs":[{"id":"measurements","kind":"query","path":"input/mini.sqlite","formats":["sqlite","sqlite3"],"required":true}],"results":{"path":"results"},"parameters":[],"requirements":["pandas"]}''')
ctx.display_parameters()
```

```python
summary = await ctx.query("measurements", "SELECT category, COUNT(*) AS count, AVG(value) AS mean_value FROM measurements GROUP BY category ORDER BY category")
summary.to_csv(ctx.results / "sqlite-summary.csv", index=False)
summary
```
